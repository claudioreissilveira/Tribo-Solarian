## **Parte 1, Altura em função do tempo (função quadrática)**

Mostra a altura do módulo depois que os retrofoguetes acendem, até o toque no solo.

$$h(t) = h_0 + v_0\,t + \frac{a\,t^2}{2}$$

| Parâmetro | Significado | Valor |
|---|---|---|
| $h_0$ | altura em que os retrofoguetes acendem | 1.800 m |
| $v_0$ | velocidade nesse momento (negativa, descendo) | -100 m/s |
| $a$ | frenagem dos retrofoguetes | 2,78 m/s² |
| $t$ | tempo desde que os retrofoguetes acendem | s |

![Figura 1](https://github.com/claudioreissilveira/Tribo-Solarian/blob/pouso/modelagem/graficos/MGPEB_altura.png?raw=true)

*Figura 1, altura em função do tempo.*

**Análise do gráfico**
- A curva é uma parábola com concavidade para cima. O vértice, em 36 s, é o toque no solo com velocidade zero.
- Com mais velocidade de chegada, os retrofoguetes precisam acender bem mais alto: dobrar a velocidade exige 4 vezes a altura.
- Com mais frenagem, eles acendem mais perto do solo e o toque acontece antes.

![Figura 2](https://github.com/claudioreissilveira/Tribo-Solarian/blob/pouso/modelagem/graficos/MGPEB_altura_sensibilidade.png?raw=true)

*Figura 2, efeito da velocidade de chegada e da frenagem.*

**Decisão no MGPEB:** os retrofoguetes acendem a 1.800 m. Como cada descida ocupa a área por 36 s, só um módulo pousa por rodada.

## **Parte 2, Combustível em função da variação de velocidade (função exponencial)**

Mostra quanto combustível o módulo gasta para frear até pousar.

$$m_c = m_f\left(e^{\Delta v / v_e} - 1\right)$$

| Parâmetro | Significado | Valor |
|---|---|---|
| $m_c$ | combustível necessário | kg |
| $m_f$ | massa do módulo | de 4.500 a 11.000 kg |
| $\Delta v$ | velocidade que os retrofoguetes precisam tirar, somando a gravidade de Marte durante a frenagem | 234,28 m/s |
| $v_e$ | velocidade de saída dos gases do motor | cerca de 3.060 m/s |

![Figura 3](https://github.com/claudioreissilveira/Tribo-Solarian/blob/pouso/modelagem/graficos/MGPEB_combustivel.png?raw=true)

*Figura 3, combustível necessário para cada módulo.*

**Análise do gráfico**
- Quanto maior o $\Delta v$, mais combustível, e cada m/s a mais custa mais que o anterior. A reta pontilhada mostra como seria se o crescimento fosse linear.
- Quanto maior a massa, mais combustível, na mesma proporção. O LOG-01 gasta 875 kg e o MED-01, 358 kg.

![Figura 4](https://github.com/claudioreissilveira/Tribo-Solarian/blob/pouso/modelagem/graficos/MGPEB_combustivel_sensibilidade.png?raw=true)

*Figura 4, uso do tanque e efeito da massa.*

**Decisão no MGPEB:** o pouso usa 19,9% do tanque em qualquer módulo. Com 30% no tanque, o módulo ainda cobre até 347 m/s, o que dá margem de segurança. Por isso o limiar vermelho do combustível é 30% e o verde é 40%.

## **Parte 3, Potência solar em função da hora do dia (função quadrática)**

Mostra quanta energia os painéis da base geram ao longo do dia marciano.

$$P(t) = P_{max}\left[1 - \left(\frac{t - D/2}{D/2}\right)^2\right] \qquad P_{max} = G \cdot A \cdot \eta \cdot e^{-\tau}$$

| Parâmetro | Significado | Valor |
|---|---|---|
| $t$ | horas desde o nascer do Sol | de 0 a 12,33 h |
| $D$ | duração do dia claro | 12,33 h |
| $G$ | luz do Sol que chega a Marte | 586,2 W/m² |
| $A$ e $\eta$ | área e eficiência dos painéis | 100 m² e 29,5% |
| $\tau$ | quantidade de poeira no ar | 0,5 no cenário calmo |

![Figura 5](https://github.com/claudioreissilveira/Tribo-Solarian/blob/pouso/modelagem/graficos/MGPEB_energia.png?raw=true)

*Figura 5, potência solar ao longo do dia.*

**Análise do gráfico**
- A curva é uma parábola com concavidade para baixo: zero no nascer e no pôr do Sol e máxima ao meio-dia, com 10,49 kW.
- A área sob a curva é a energia do dia: 86,2 kWh no cenário calmo.
- Com mais poeira, a parábola inteira encolhe. O pico cai de forma exponencial: com $\tau = 3$ ele fica em só 0,86 kW.

![Figura 6](https://github.com/claudioreissilveira/Tribo-Solarian/blob/pouso/modelagem/graficos/MGPEB_energia_sensibilidade.png?raw=true)

*Figura 6, efeito da poeira na geração de energia.*

**Decisão no MGPEB:** com poeira no limiar vermelho ($\tau = 3$) a base gera só 7,1 kWh por dia, por isso o pouso é adiado. O módulo de energia pousa primeiro para a geração começar antes dos outros.

## **Parte 4, Resumo**

| Modelo | Função | Valor principal | Decisão no MGPEB |
|---|---|---|---|
| Altura × tempo | quadrática | toque em 36 s, retrofoguetes a 1.800 m | quando acender os retrofoguetes; um pouso por rodada |
| Combustível × Δv | exponencial | Δv de 234,28 m/s, 19,9% do tanque | limiares de combustível de 30% e 40% |
| Potência × hora | quadrática | pico de 10,49 kW e 86,2 kWh por dia | limiares de poeira; módulo de energia pousa primeiro |
